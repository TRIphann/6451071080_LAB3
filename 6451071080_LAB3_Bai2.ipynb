```python
print("Họ tên: Phan Công Trí - MSSV: 6451071080")

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# a) Đọc bộ dữ liệu train, test của MNIST từ Keras
(train_images, train_labels), (test_images, test_labels) = mnist.load_data()
```

```python
# ==================== CÂU a, b, c, d ====================
print("Họ tên: Phan Công Trí - MSSV: 6451071080")

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# b) In ra kích thước của tập train, test
print("Kích thước train_images ban đầu:", train_images.shape)
print("Kích thước train_labels ban đầu:", train_labels.shape)
print("Kích thước test_images:", test_images.shape)
print("Kích thước test_labels:", test_labels.shape)
```

```python
# ==================== CÂU a, b, c, d ====================
print("Họ tên: Phan Công Trí - MSSV: 6451071080")

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# c) Chuẩn hóa các giá trị pixel của ảnh về đoạn [0,1]
train_images = train_images / 255.0
test_images = test_images / 255.0

# d) Chia tập train ban đầu thành tập train và tập validation theo tỷ lệ 8:2
train_images, validation_images, train_labels, validation_labels = train_test_split(
    train_images, train_labels, test_size=0.2, random_state=42
)

print("\nSau khi chia 8:2:")
print("Kích thước train_images:", train_images.shape, "| train_labels:", train_labels.shape)
print("Kích thước validation_images:", validation_images.shape, "| validation_labels:", validation_labels.shape)
```

```python
# ==================== CÂU e, f, g ====================
print("Họ tên: Phan Công Trí - MSSV: 6451071080")

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Flatten, Dense, Dropout

# e) Xây dựng Neural Network
model = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(128, activation='relu'),
    Dropout(0.2),
    Dense(10, activation='softmax')
])

# f) Biên dịch mô hình
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

model.summary()

# g) Huấn luyện mô hình: epoch=5, batch_size=32
history = model.fit(
    train_images, train_labels,
    validation_data=(validation_images, validation_labels),
    epochs=5,
    batch_size=32
)
```

```python
# ==================== CÂU h ====================
print("Họ tên: Phan Công Trí - MSSV: 6451071080")

# h) Vẽ biểu đồ thể hiện accuracy, loss sau mỗi epoch
epochs_range = range(1, 6)
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(epochs_range, history.history['accuracy'], marker='o', label='Train Accuracy')
plt.plot(epochs_range, history.history['val_accuracy'], marker='o', label='Validation Accuracy')
plt.title('Accuracy qua các Epoch')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(epochs_range, history.history['loss'], marker='o', label='Train Loss')
plt.plot(epochs_range, history.history['val_loss'], marker='o', label='Validation Loss')
plt.title('Loss qua các Epoch')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# ==================== CÂU i, j, k, l ====================
print("Họ tên: Phan Công Trí - MSSV: 6451071080")

# i) Dự đoán phân lớp cho tập test
y_prob = model.predict(test_images)
y_pred = np.argmax(y_prob, axis=1)

# j) Tính và in ra Accuracy của mô hình trên tập test
test_acc = accuracy_score(test_labels, y_pred)
print("Accuracy trên tập test:", test_acc)

# k) Tính và in ra Precision, Recall, F1-score của từng lớp và trung bình
print("\nClassification Report trên tập test:\n")
print(classification_report(test_labels, y_pred, digits=4))

# l) Hiển thị confusion matrix bằng heat map
cm = confusion_matrix(test_labels, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.title('Confusion Matrix - MNIST (Train/Val/Test)')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.show()
```

```python
# ==================== CÂU m, n ====================
print("Họ tên: Phan Công Trí - MSSV: 6451071080")

# m) In ra kết quả phân lớp của 05 ảnh đầu tiên trong tập test
plt.figure(figsize=(12, 3))
for i in range(5):
    img = np.expand_dims(test_images[i], axis=0)
    pred_label = np.argmax(model.predict(img, verbose=0), axis=1)[0]

    plt.subplot(1, 5, i + 1)
    plt.imshow(test_images[i], cmap='gray')
    plt.title(f"Pred: {pred_label} | True: {test_labels[i]}")
    plt.axis('off')
plt.suptitle("05 ảnh đầu tiên trong tập Test", fontsize=14)
plt.tight_layout()
plt.show()

# n) In ra 05 kết quả đầu tiên phân lớp sai so với nhãn thực tế trong tập test
misclassified_idx = np.where(y_pred != test_labels)[0][:5]
print("Chỉ số 5 mẫu phân lớp sai đầu tiên:", misclassified_idx)

plt.figure(figsize=(12, 3))
for idx, wrong_i in enumerate(misclassified_idx):
    plt.subplot(1, 5, idx + 1)
    plt.imshow(test_images[wrong_i], cmap='gray')
    plt.title(f"Idx {wrong_i}: Pred {y_pred[wrong_i]} (True {test_labels[wrong_i]})", color='red')
    plt.axis('off')
plt.suptitle("05 ảnh đầu tiên phân lớp SAI trong tập Test", fontsize=14)
plt.tight_layout()
plt.show()
```
