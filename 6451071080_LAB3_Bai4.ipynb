```python
# ==================== CÂU a, b, c ====================
print("Họ tên: Phan Công Trí - MSSV: 6451071080")

import numpy as np
import tensorflow as tf
from sklearn.datasets import load_iris
from sklearn.model_selection import KFold
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# a) Đọc bộ dữ liệu Iris từ sklearn
iris = load_iris()
X = iris.data
Y = iris.target

# b) Chuẩn hóa dữ liệu data về đoạn [0,1]
X_max = X.max(axis=0)
X = X / X_max

# c) Định nghĩa hàm create_model (tương tự câu 1.h)
def create_model():
    model = Sequential([
        Dense(10, activation='relu', input_shape=(X.shape[1],)),
        Dense(20, activation='relu'),
        Dense(len(iris.target_names), activation='softmax')
    ])
    model.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])
    return model

print("Đã chuẩn bị dữ liệu và hàm create_model() thành công!")
```

```python
# ==================== CÂU d ====================
print("Họ tên: Phan Công Trí - MSSV: 6451071080")

# d) Huấn luyện mô hình bằng k-fold cross validation (k=10)
kfold = KFold(n_splits=10, shuffle=True, random_state=42)

models_iris = []
acc_folds = []
prec_folds = []
rec_folds = []
f1_folds = []

prec_per_class = []
rec_per_class = []
f1_per_class = []

fold_no = 1
for train_idx, val_idx in kfold.split(X):
    model = create_model()
    X_train_f, X_val_f = X[train_idx], X[val_idx]
    y_train_f, y_val_f = Y[train_idx], Y[val_idx]

    # Huấn luyện
    model.fit(X_train_f, y_train_f, epochs=100, batch_size=1, verbose=0)

    # Dự đoán trên tập validation của fold
    y_pred_f = np.argmax(model.predict(X_val_f, verbose=0), axis=1)

    # Độ đo trung bình của fold
    acc = accuracy_score(y_val_f, y_pred_f)
    prec = precision_score(y_val_f, y_pred_f, average='macro', zero_division=0)
    rec = recall_score(y_val_f, y_pred_f, average='macro', zero_division=0)
    f1 = f1_score(y_val_f, y_pred_f, average='macro', zero_division=0)

    acc_folds.append(acc)
    prec_folds.append(prec)
    rec_folds.append(rec)
    f1_folds.append(f1)

    # Độ đo cho từng class (0, 1, 2)
    prec_per_class.append(precision_score(y_val_f, y_pred_f, labels=[0, 1, 2], average=None, zero_division=0))
    rec_per_class.append(recall_score(y_val_f, y_pred_f, labels=[0, 1, 2], average=None, zero_division=0))
    f1_per_class.append(f1_score(y_val_f, y_pred_f, labels=[0, 1, 2], average=None, zero_division=0))

    models_iris.append(model)
    print(f"Fold {fold_no:2d} | Accuracy: {acc:.4f} | Precision: {prec:.4f} | Recall: {rec:.4f} | F1-score: {f1:.4f}")
    fold_no += 1
```

```python
# ==================== CÂU e, f, g ====================
print("Họ tên: Phan Công Trí - MSSV: 6451071080")

# e) In ra Average Accuracy, Average Recall, Average Precision, Average F1-score
print("=== KẾT QUẢ TRUNG BÌNH TRÊN TOÀN BỘ 10 FOLDS ===")
print(f"Average Accuracy  : {np.mean(acc_folds):.4f}")
print(f"Average Precision : {np.mean(prec_folds):.4f}")
print(f"Average Recall    : {np.mean(rec_folds):.4f}")
print(f"Average F1-score  : {np.mean(f1_folds):.4f}")

# f) In ra Average Recall, Average Precision, Average F1-score cho từng class
avg_prec_class = np.mean(prec_per_class, axis=0)
avg_rec_class = np.mean(rec_per_class, axis=0)
avg_f1_class = np.mean(f1_per_class, axis=0)

print("\n=== KẾT QUẢ TRUNG BÌNH CHO TỪNG CLASS ===")
for i, class_name in enumerate(iris.target_names):
    print(f"Lớp {class_name:10s} -> Avg Precision: {avg_prec_class[i]:.4f} | Avg Recall: {avg_rec_class[i]:.4f} | Avg F1-score: {avg_f1_class[i]:.4f}")

# g) Sử dụng mô hình có F1-score tốt nhất dự đoán cho 03 mẫu mới
best_f1_idx = np.argmax(f1_folds)
best_f1_model = models_iris[best_f1_idx]
print(f"\nMô hình có F1-score cao nhất là ở Fold {best_f1_idx + 1} (F1 = {f1_folds[best_f1_idx]:.4f})")

new_samples = np.array([
    [6.2, 2.9, 4.3, 1.3],
    [5.1, 3.5, 1.4, 0.2],
    [7.3, 2.8, 6.4, 2.1]
])
new_samples_norm = new_samples / X_max

preds_new = np.argmax(best_f1_model.predict(new_samples_norm, verbose=0), axis=1)
for idx, sample in enumerate(new_samples):
    print(f"Mẫu {sample} -> Dự đoán: {iris.target_names[preds_new[idx]]}")
```
